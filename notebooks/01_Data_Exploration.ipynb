## Outlier Detection

Identify potential outliers in the numerical variables to determine whether any values appear unusual or require further investigation before data cleaning.

```python
numerical_columns = [
    "Zip Code",
    "Latitude",
    "Longitude",
    "Tenure Months",
    "Monthly Charges",
    "Churn Score",
    "CLTV"
]
```

## Summary Statistics

Review the numerical variables to identify unusually high or low values.

```python
df[numerical_columns].describe().T
```

## Box Plots

Visualize the distribution of each numerical variable and identify potential outliers.

```python
for column in numerical_columns:
    plt.figure(figsize=(8, 1.5))
    plt.boxplot(df[column].dropna(), vert=False)
    plt.title(column)
    plt.show()
```

## Initial Observations

- No significant numerical outliers requiring removal were identified.
- Monthly Charges and Tenure Months appear to have reasonable distributions.
- CLTV values are widely distributed but appear to represent legitimate customer records.
- Geographic variables (Latitude and Longitude) fall within expected ranges.
- No numerical values appear to indicate obvious data quality issues.

## Key Findings

- The dataset contains 7,043 customer records and 33 variables.
- No duplicate records were identified.
- Churn Reason contains missing values for customers who did not churn.
- Approximately 26.5% of customers churned.
- The dataset is ready for data cleaning.
